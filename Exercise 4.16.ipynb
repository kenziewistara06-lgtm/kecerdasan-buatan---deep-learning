```python
# Exercise 4.16
# Display all filters in the first convolutional layer

from keras.applications.vgg19 import VGG19
from matplotlib import pyplot as plt

# Load VGG19 model
model = VGG19()

# Select the first convolutional layer
n = 1

# Get filters and biases
filters, biases = model.layers[n].get_weights()

# Get filter shape
s = filters.shape

print("Color channels:", s[2])
print("Filter size:", s[0], s[1])
print("Total number of filters:", s[3])
print("Total number of biases:", s[3])

# Normalize filter values to 0-1
f_min, f_max = filters.min(), filters.max()
filters = (filters - f_min) / (f_max - f_min)

# Display all filters
n_filters, ix = s[3], 1

plt.figure(figsize=(10, 80))

for i in range(n_filters):
    # Get the filter
    f = filters[:, :, :, i]

    # Plot each color channel separately
    for j in range(s[2]):
        ax = plt.subplot(n_filters, s[2], ix)
        ax.set_xticks([])
        ax.set_yticks([])
        plt.imshow(f[:, :, j], cmap='gray')
        ix += 1

plt.tight_layout()
plt.show()
```
