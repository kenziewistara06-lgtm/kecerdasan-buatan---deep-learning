```python
# Exercise 4.15
# Visualize filters from the second convolutional layer

from keras.applications.vgg19 import VGG19
from matplotlib import pyplot as plt


# ============================================================
# 1. LOAD VGG19 MODEL
# ============================================================

model = VGG19()


# ============================================================
# 2. SELECT THE SECOND CONVOLUTIONAL LAYER
# ============================================================

n = 2

filters, biases = model.layers[n].get_weights()


# ============================================================
# 3. GET FILTER SHAPE
# ============================================================

s = filters.shape

print("Layer:", model.layers[n].name)
print("Color channels:", s[2])
print("Filter size:", s[0], s[1])
print("Total number of filters:", s[3])
print("Total number of biases:", s[3])


# ============================================================
# 4. NORMALIZE FILTER VALUES
# ============================================================

f_min, f_max = filters.min(), filters.max()

filters = (filters - f_min) / (f_max - f_min)


# ============================================================
# 5. DISPLAY FIRST FOUR FILTERS
# ============================================================

n_filters, ix = 4, 1

plt.figure(figsize=(10, 10))

for i in range(n_filters):

    # Get the filter
    f = filters[:, :, :, i]

    # Display each input channel
    for j in range(s[2]):

        ax = plt.subplot(n_filters, s[2], ix)

        ax.set_xticks([])
        ax.set_yticks([])

        plt.imshow(
            f[:, :, j],
            cmap='gray'
        )

        ix += 1


plt.tight_layout()
plt.show()
```
