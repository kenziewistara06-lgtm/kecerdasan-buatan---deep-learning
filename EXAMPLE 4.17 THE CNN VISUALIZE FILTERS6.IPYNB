```python
# Example 4.17 CNN Visualize Filters.ipynb
# Modified from:
# https://machinelearningmastery.com/how-to-visualize-filters-and-feature-maps-in-convolutional-neural-networks/


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

from keras.applications.vgg19 import VGG19, preprocess_input
from keras.preprocessing.image import load_img, img_to_array
from keras.models import Model
from matplotlib import pyplot as plt
from numpy import expand_dims


# ============================================================
# 2. LOAD VGG19 MODEL
# ============================================================

model = VGG19()

# Display model architecture
model.summary()


# ============================================================
# 3. DISPLAY LAYER NUMBERS AND LAYER NAMES
# ============================================================

print("\nLayer numbers and names:")

n = 0

for layer in model.layers:
    print(n, layer.name)
    n += 1


# ============================================================
# 4. DISPLAY FILTER AND BIAS SHAPES
#    FOR CONVOLUTIONAL LAYERS
# ============================================================

print("\nConvolutional layer filters and biases:")

n = 0

for layer in model.layers:

    if 'conv' in layer.name:
        filters, biases = layer.get_weights()

        print(
            n,
            layer.name,
            filters.shape,
            biases.shape
        )

    n += 1


# ============================================================
# 5. VISUALIZE FILTERS FROM THE FIRST CONVOLUTIONAL LAYER
# ============================================================

# Select the first convolutional layer
n = 1

filters, biases = model.layers[n].get_weights()

# Get filter shape
s = filters.shape

print("\nFilter information:")
print("Color channels:", s[0])
print("Filter size:", s[1], s[2])
print("Total number of filters:", s[3])


# Normalize filter values to 0-1
# so they can be visualized
f_min, f_max = filters.min(), filters.max()

filters = (filters - f_min) / (f_max - f_min)


# Plot the first four filters
n_filters = 4
ix = 1

plt.figure(figsize=(10, 10))

for i in range(n_filters):

    # Get the filter
    f = filters[:, :, :, i]

    # Plot each color channel separately
    for j in range(s[0]):

        ax = plt.subplot(
            n_filters,
            s[0],
            ix
        )

        ax.set_xticks([])
        ax.set_yticks([])

        plt.imshow(
            f[j, :, :],
            cmap='gray'
        )

        ix += 1


# Show the filters
plt.show()


# ============================================================
# 6. FUNCTION TO PLOT FEATURE MAPS
# ============================================================

def plot_feature_maps(feature_maps):

    # Number of columns
    col = 8

    # Calculate number of rows
    row = int(feature_maps.shape[3] / col)

    ix = 1

    plt.figure(figsize=(20, 20))

    for _ in range(row):

        for _ in range(col):

            # Specify subplot
            ax = plt.subplot(
                row,
                col,
                ix
            )

            # Turn off axis
            ax.set_xticks([])
            ax.set_yticks([])

            # Plot feature map in grayscale
            plt.imshow(
                feature_maps[0, :, :, ix - 1],
                cmap='gray'
            )

            ix += 1

    # Show feature maps
    plt.show()


# ============================================================
# 7. CREATE MODEL FOR FEATURE MAP VISUALIZATION
# ============================================================

# Load a new VGG19 model
model = VGG19()

# Select the hidden layer to visualize
n = 1

# Redefine the model to output
# the selected hidden layer
feature_model = Model(
    inputs=model.inputs,
    outputs=model.layers[n].output
)

feature_model.summary()


# ============================================================
# 8. LOAD IMAGE
# ============================================================

img = load_img(
    'Elephant.jpg',
    target_size=(224, 224)
)


# ============================================================
# 9. CONVERT IMAGE TO ARRAY
# ============================================================

img = img_to_array(img)


# ============================================================
# 10. EXPAND DIMENSIONS
# ============================================================

img = expand_dims(
    img,
    axis=0
)


# ============================================================
# 11. PREPROCESS IMAGE FOR VGG19
# ============================================================

img = preprocess_input(img)


# ============================================================
# 12. GET FEATURE MAPS
# ============================================================

feature_maps = feature_model.predict(
    img,
    verbose=0
)

print(
    "\nFeature maps:",
    feature_maps.shape
)


# ============================================================
# 13. VISUALIZE FEATURE MAPS
# ============================================================

plot_feature_maps(feature_maps)
```
